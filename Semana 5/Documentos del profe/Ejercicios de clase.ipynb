```python
# D6 🔴 El ticket del merch
precios = {"playera": 350, "gorra": 220, "parche": 80, "sudadera": 650}
carrito = {"playera": 2, "parche": 3}
total = 0

for producto, cantidad in carrito.items():
    total += precios[producto] * cantidad

print(f"Total: ${total}")
```

```python
def seccion_tuplas():
    print("Aqui van las tuplas")

def seccion_strings():
    print("Aqui van los strings")

while True:
    op = input("Opción (1-3): ")
    if op == "1":
        seccion_tuplas()
    elif op == "2":
        seccion_strings()
    elif op == "3":
        break
    else:
        print("Opción inválida")
```
