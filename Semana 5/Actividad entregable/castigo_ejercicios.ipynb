# Ejercicio 1 — Bitácora de la semana

```python
dias = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

def resumen(km):
    total = 0
    dias_rodados = 0
    indice_mayor = 0

    for i in range(len(km)):
        total += km[i]

        if km[i] >0:
            dias_rodados += 1

        if km[i] > km[indice_mayor]:
            indice_mayor = i

    promedio = round(total / dias_rodados, 1)

    return total, dias_rodados, promedio, indice_mayor

# Desempacar resultados
total, cantidad_dias, promedio, indice = resumen(km)

print("Total:", total, "km")
print("Días con rodada:", cantidad_dias)
print("Promedio por día rodado:", promedio, "km")
print("Día más largo:", dias[indice], f"{km[indice]} km")
```

# Ejercicio 3 — Tramos de la ruta a la Sierra

```python
paradas = (("Queretaro", 0), ("SanJuan del Río", 52), ("Tequisquiapan", 74), ("Bernal", 125), ("Jalpan", 260))

tramo_mayor = 0

for i in range(len(paradas) - 1):

    origen = paradas[i][0]
    km_origen = paradas[i][1]

    destino = paradas[i +1][0]
    km_destino = paradas[i + 1][1]

    distancia = km_destino - km_origen

    print(f"{origen} -> {destino}: {distancia} km")

    if distancia > tramo_mayor:
        tramo_mayor = distancia
        parada_inicial = distancia
        parada_final = destino

print(f"Tramo más largo: {parada_inicial} -> {parada_final} ({tramo_mayor} km)")
```

# Ejercicio 4 — ¿Con chamarra o sin chamarra?

```python
temps = (8, 14, 22, 27, 31, 19, 11)

# Clasificar las temperaturas
def clasificar(t):
    if t < 12:
        return "frío"
    elif t <= 24:
        return "templado"
    else:
        return "calor"

# Lista 
clasificaciones = []

for t in temps:
    resultado = clasificar(t)
    clasificaciones.append(resultado)

# Diccionario contador
contador = {"frío": 0, "templado": 0, "calor": 0}

for categoria in clasificaciones:
    contador[categoria] += 1


# Categoría que predomina
predomina = ""
mayor = 0

for categoria, cantidad in contador.items():
    if cantidad > mayor:
        mayor = cantidad
        predomina = categoria

# Resultados
print(clasificaciones)
print(contador)
print("Predomina:", predomina)
```

# Ejercicio 5 — ¿Cuánto dura el show?

```python
setlist = (("Intro", 95), ("Riff asesino", 245), ("Balada", 302), ("Solo", 128), ("Encore", 409))

def formato(seg):
    minutos = seg // 60
    segundos = seg % 60

    if segundos < 10:
        return f"f{minutos}:0{segundos}"
    else:
        return f"{minutos}:{segundos}"

total = 0

for cancion, duracion in setlist:
    print(f"{cancion}: {formato(duracion)}")
    total += duracion

print("Duración total:", formato(total))
```

# Ejercicio 7 — Boleta del grupo


```python
notas = {"Ana": [90, 85, 78], "Luis": [60, 72, 65], "Mía": [100, 95, 98], "Beto": [70, 68, 74]}

def promedio(valores):
    acumulado = 0

    for numero in valores:
        acumulado = acumulado + numero

    resultado = acumulado / len(valores)
    return round(resultado, 1)

resultados = {}

for estudiante, lista_notas in notas.items():
    resultados[estudiante] = promedio(lista_notas)

for estudiante, nota_final in resultados.items():
    estado = "Reprobado"

    if nota_final >= 70:
        estado = "Aprobado"

    print(f"{estudiante}: {nota_final} - {estado}")

destacado = max(resultados, key=resultados.get)

print("Mejor promedio:", destacado)
```

# Ejercicio 10 — Traductor de mensajes de WhatsApp


```python
abrev = {"q": "que", "xq": "porque", "tmb": "también", "pq": "porque", "bn": "bien", "tqm": "te quiero mucho"}

def expandir(mensaje):
    texto = mensaje.lower()
    palabras = texto.split()

    traduccion = [
        abrev.get(palabra, palabra)
        for palabra in palabras
    ]

    return " ".join(traduccion)

mensajes = [
    "Q onda tmb voy al concierto xq tocan bn",
    "tqm Mamá"
]

for texto in mensajes:
    print(expandir(texto))
```

# Ejercicio 11 — Rendimiento de la moto



```python
def rendimiento():
    while True:
        try:
            distancia = float(input("Km recorridos: "))
            combustible = float(input("Litros utilizados: "))

            if combustible == 0:
                raise ZeroDivisionError

            consumo = distancia / combustible
            return round(consumo, 1)

        except ValueError:
            print("Escribe solo números")

        except ZeroDivisionError:
            print("Los litros no pueden ser 0")

resultado = rendimiento()

print(f"Rendimiento: {resultado} km/l")

if resultado < 15:
    print("Revisa tu moto")
elif resultado < 20:
    print("Normal")
else:
    print("Excelente")
```

# Ejercicio 12 — Taquilla familiar

```python
def precio_boleto(anios):
    if anios >= 60:
        return 180
    elif anios >= 12:
        return 300
    else:
        return 150

cuenta = 0

for turno in range(1, 4):
    while True:
        try:
            anios = int(input(f"Edad de la persona {turno}: "))

            if 0 <= anios <= 120:
                break
            else:
                print("Edad fuera de rango")

        except ValueError:
            print("Escribe la edad con número")

    costo = precio_boleto(anios)
    cuenta = cuenta + costo

    print(f"Boleto: ${costo}")

print(f"Total a pagar: ${cuenta}")
```

# Ejercicio 13 — Capturas sucias


```python
capturas = ["10", "7", "x", "", "9", "8.5", " 6 "]

numeros_correctos = []
errores = 0
suma = 0

for registro in capturas:
    try:
        calificacion = int(registro)
    except ValueError:
        errores = errores + 1
    else:
        numeros_correctos.append(calificacion)
        suma = suma + calificacion

media = round(suma / len(numeros_correctos), 2)

print("Válidas:", numeros_correctos)
print("Inválidas:", errores)
print("Promedio:", media)
```

" 6 " sí se pudo convertir y "8.5" no porque int() permite espacios al inicio y al final de una cadena, por eso " 6 " se convierte correctamente en 6. Y "8.5" genera un error porque int() solo acepta números enteros escritos como texto, no decimales.

# Ejercicio 14 — Reserva tu asiento

```python
sala = [["X", "L", "L"], ["L", "X", "L"], ["X", "X", "L"]]

def reservar(matriz):
    try:
        renglon = int(input("Fila: "))
        lugar = int(input("Columna: "))

        if renglon < 0 or lugar < 0:
            raise IndexError

        estado = matriz[renglon][lugar]

        if estado == "L":
            matriz[renglon][lugar] = "X"
            print("¡Reservado!")
            return True

        print("Ocupado, elige otro")
        return False

    except ValueError:
        print("Escribe números")
        return False

    except IndexError:
        print("Ese asiento no existe")
        return False

reserva_lista = False

while reserva_lista == False:
    reserva_lista = reservar(sala)

for renglon in sala:
    for lugar in renglon:
        print(lugar, end=" ")
    print()
```

# Ejercicio 15 — Cupones de descuento


```python
cupones = {"METAL10": 10, "MOTO20": 20, "ROCK15": 15}

def aplicar(cantidad, clave):
    codigo_limpio = clave.strip().upper()

    try:
        porcentaje = cupones[codigo_limpio]
    except KeyError:
        print("Cupón no válido")
        return cantidad

    ahorro = cantidad * porcentaje / 100
    return cantidad - ahorro

pruebas = ["moto20 ", "GRATIS", "metal10"]

for clave in pruebas:
    precio_nuevo = aplicar(1200, clave)
    print(f"'{clave}' -> ${precio_nuevo}")
```

# Ejercicio 17 — Se lee igual al revés

```python
def es_palindromo(texto):
    sin_espacios = texto.replace(" ", "")
    minusculas = sin_espacios.lower()

    invertida = minusculas[::-1]

    return minusculas == invertida

textos = ("Anita lava la tina", "ABBA", "Metallica", "Oso")

for elemento in textos:
    respuesta = es_palindromo(elemento)
    print(f"{elemento}: {respuesta}")
```

# Ejercicio 18 — Contador de vocales


```python
frase = "el rock nunca va a morir en esta ciudad"

def vocales(palabra):
    cantidad = 0

    for caracter in palabra:
        if caracter in "aeiou":
            cantidad = cantidad + 1

    return cantidad

conteo = {}

for termino in frase.split():
    conteo[termino] = vocales(termino)

print(conteo)

ganadora = max(conteo, key=conteo.get)

print(f"Más vocales: {ganadora} ({conteo[ganadora]})")
```

# Ejercicio 19 — Usuarios del laboratorio

```python
alumnos = ["Bruce Dickinson", "Brian Dickinson", "Ana Lopez", "Andrea Lopez", "Angel Lopez"]

registrados = []

def usuario(nombre):
    datos = nombre.split()
    inicial = datos[0][0]
    apellido = datos[1]

    return (inicial + apellido).lower()

for persona in alumnos:
    identificador = usuario(persona)
    disponible = identificador
    consecutivo = 1

    while disponible in registrados:
        consecutivo += 1
        disponible = identificador + str(consecutivo)

    registrados.append(disponible)

    print(f"{persona} → {disponible}")
```

# Ejercicio 20 — El censor del salón


```python
prohibidas = ["tonto", "menso", "baboso"]

def censurar(frase):
    contenido = frase.split()
    reemplazos = 0

    for posicion in range(len(contenido)):
        termino = contenido[posicion]

        if termino.lower() in prohibidas:
            contenido[posicion] = "*" * len(termino)
            reemplazos += 1

    mensaje_nuevo = " ".join(contenido)

    return mensaje_nuevo, reemplazos

ejemplos = ["no seas menso y pásame el código", "Tonto el que no haga commit",
 "todo bien profe"]

for ejemplo in ejemplos:
    mensaje_final, total_censuradas = censurar(ejemplo)

    print(f"{mensaje_final}  ({total_censuradas} censuradas)")
```
