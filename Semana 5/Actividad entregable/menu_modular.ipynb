El programa utiliza un menú para realizar diferentes actividades con tuplas, diccionarios, operaciones y strings. También permite buscar contactos y manejar errores cuando se ingresan datos incorrectos, además de ofrecer una opción para terminar la ejecución. 

```python
# PARTE A: TUPLAS

def calcular_total(elementos):
    acumulado = 0

    for dato in elementos:
        acumulado = acumulado + dato
 
    return acumulado


def mostrar_tuplas():

    numeros = (27, 24, 17, 3, 2)

    print("\nTupla original:", numeros)
    print("Tercer elemento:", numeros[2])

    primer_numero = float(input("Escribe un numero: "))
    segundo_numero = float(input("Escribe otro numero: "))

    nueva_tupla = numeros + (primer_numero, segundo_numero)

    print("Tupla nueva:", nueva_tupla)

    ordenados = list(nueva_tupla)
    ordenados.sort()

    print("Numeros ordenados:", ordenados)

    suma_final = calcular_total(nueva_tupla)

    print("Resultado de la suma:", suma_final)


# PARTE B: DICCIONARIOS

def obtener_telefono(diccionario, persona):

    if persona in diccionario:
        return diccionario[persona]
    else:
        return None


def administrar_contactos():

    contactos = {"Fernanda": "442-1153", "Regina": "446-7351", "Sofia": "442-0321"}

    persona_nueva = input("Nombre del nuevo contacto: ")
    numero_nuevo = input("Telefono del contacto: ")

    contactos[persona_nueva] = numero_nuevo

    print("\nLista de nombres registrados:")

    for persona in contactos.keys():
        print(persona)

    consulta = input("\nEscribe el nombre que deseas buscar: ")

    telefono_encontrado = obtener_telefono(contactos, consulta)

    if telefono_encontrado is not None:
        print("Numero telefonico:", telefono_encontrado)
    else:
        print(consulta, "no esta en tus contactos.")


# PARTE C: EXCEPCIONES

def operaciones_seguras():

    try:
        primer_valor = int(input("Ingresa el primer numero entero: "))
        segundo_valor = int(input("Ingresa el segundo numero entero: "))

        resultado_suma = primer_valor + segundo_valor
        print("Resultado de la suma:", resultado_suma)

        resultado_division = primer_valor / segundo_valor
        print("Resultado de la division:", resultado_division)

    except ValueError:
        print("Dato incorrecto. Debes ingresar numeros enteros.")

    except ZeroDivisionError:
        print("No se puede realizar una division entre cero.")


# PARTE D: STRINGS

def cantidad_palabras(texto):

    lista_palabras = texto.split()

    return len(lista_palabras)


def analizar_texto():

    mensaje = "Espero me vaya bien en esta actividad"

    print("\nTexto original:", mensaje)
    print("Cantidad de caracteres:", len(mensaje))
    print("Texto en mayusculas:", mensaje.upper())

    nuevo_mensaje = mensaje.replace("actividad", "tarea")

    print("Texto modificado:", nuevo_mensaje)

    palabras_en_mensaje = cantidad_palabras(mensaje)

    print("Cantidad de palabras:", palabras_en_mensaje)


# PARTE E: MENU PRINCIPAL

def iniciar_programa():

    opcion = ""

    while opcion != "5":

        print("\n===== MENU PRINCIPAL =====")
        print("1. Trabajar con tuplas")
        print("2. Administrar contactos")
        print("3. Realizar operaciones")
        print("4. Analizar strings")
        print("5. Salir")

        opcion = input("Selecciona una opcion: ")

        if opcion == "1":
            mostrar_tuplas()

        elif opcion == "2":
            administrar_contactos()

        elif opcion == "3":
            operaciones_seguras()

        elif opcion == "4":
            analizar_texto()

        elif opcion == "5":
            print("El programa ha terminado.")

        else:
            print("Opcion invalida. Intenta nuevamente.")


iniciar_programa()
```
