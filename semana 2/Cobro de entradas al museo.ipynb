```python
#Actividad 2 taquilla
#Fernanda Yvonne Juarez Plaza
# Matricula: 3084460
# El programa consiste en analizar los visitantes y sus diferentes tipos así como sus edades y poderlos clasificar, al igual de tener una total de la información y que sea claro y poder ayudar al museo a controlar errores y carga administrativa





# Mensaje inicial del programa
print("====== Taquilla del museo de antropología e historia ======")

# Precios
PRECIO1 = 0
PRECIO2 = 30
PRECIO3 = 45

# Descuentos
DESCUENTO_MAYOR = 0.12
DESCUENTO_ESTUDIANTE = 0.10
DESCUENTO_PROFESOR = 0.10

# Contador y acumuladores
cantvisit = 0
total = 0
total_descuentos = 0

# Ciclo de visitantes
while True:

    entrada = input("Edad del visitante (o 'fin' para cerrar): ")

    if entrada == "fin":
        break

    edad = int(entrada)

    # Validación de edad
    if edad <= 0 or edad >= 110:
        print("Edad no válida")
        continue

    # Precio según edad
    if edad < 3:
        precio = PRECIO1
    elif edad <= 17:
        precio = PRECIO2
    else:
        precio = PRECIO3

    # Descuentos
    descuento = 0
    tipo = "No aplica"

    if edad >= 18:

        opcion = input("Tipo: 1) Estudiante  2) Profesor  3) Ninguno: ")

        if edad >= 60:
            tipo = "Adulto mayor"
            descuento = precio * DESCUENTO_MAYOR
        elif opcion == "1":
            tipo = "Estudiante"
            descuento = precio * DESCUENTO_ESTUDIANTE
        elif opcion == "2":
            tipo = "Profesor"
            descuento = precio * DESCUENTO_PROFESOR
        else:
            tipo = "Ninguno"

    precio_final = precio - descuento

    # Acumuladores y contador
    cantvisit = cantvisit + 1
    total = total + precio_final
    total_descuentos = total_descuentos + descuento

    # Desglose del visitante
    print(f"Visitante {cantvisit} | Edad: {edad} | {tipo} | Base: ${precio:.2f} | Descuento: ${descuento:.2f} | A pagar: ${precio_final:.2f}")

# Corte de caja
print("====== CORTE DE CAJA ======")
print("Visitantes cobrados:", cantvisit)
print(f"Total en descuentos: ${total_descuentos:.2f}")
print(f"TOTAL A PAGAR: ${total:.2f}")
```
