```python
#L5 🟡 Duración del disco

duraciones = [3.36, 4.11, 4.30]
for d in duraciones:
    print(f"Duracion: {d} minutos")
duraciontotal = 0
for d in duraciones:
    duraciontotal = duraciontotal + d
    print(f"Duracion total: {duraciontotal} minutos")
promedio = duraciontotal / len(duraciones)
print(f"Promedio: {promedio} minutos")
```

```python
duraciones = [3.28, 3.45, 3.44, 4.26, 3.56]

# Calculamos el total usando la función sum() de Python
duraciontotal = sum(duraciones)
promedio = duraciontotal / len(duraciones)

# Imprimimos únicamente los resultados finales
print(f"Duración total: {duraciontotal:.2f} minutos")
print(f"Promedio: {promedio:.2f} minutos")
```

```python
# M4 🟡 Total por vendedor

ventas = [[100, 200, 150], [80, 90, 120]]

for i in range(len(ventas)):
    total = sum(ventas[i])
    print(f"vendedor {i + 1}: {total}")
print("\nTotal de ventas por día: ")
for j in range(len(ventas[i])):
    total_dia = sum(ventas[i][j] for i in range(len(ventas)))
    print(f"día {j + 1}: {total_dia}")


```

```python
# M6 🔴 El gato

# M6 El gato
tablero = [["X", "O", "X"],
           [" ", "X", "O"],
           ["O", " ", "X"]]

for fila in tablero:
    print(" | ".join(fila))
    if fila != tablero[2]:
        print("---------")

if tablero[0][0] == "X" and tablero[1][1] == "X" and tablero[2][2] == "X":
    print("¡X gana por la diagonal!")

```

```python
# R1 🟢 ¿Pasa el cadenero?

def es_mayor(edad):
    return True if edad >= 18 else False

edades = [15, 22, 17, 30, 19, 12]
personas = ["Sofa", "Nicolette", "Milena", "Fernanda", "Josue", "Fanny"]

for i in range(len(edades)):
    if es_mayor(edades[i]):
        print(personas[i], edades[i], "años: pasa")
    else:
        print(personas[i], edades[i], "años: no pasa")
```

```python
# R3 🟡 Promedio por alumno

def promedio(fila):
    return sum(fila) / len(fila)
calificaciones = [[8, 9, 10], [6, 7, 8]]
for i, fila in enumerate(calificaciones, start=1):
    print(f"Alumno{i}: {promedio(fila):.2f}")
```
