Este programa crea una tabla de Pitágoras del 1 al 10 utilizando una matriz. Permite mostrar los resultados en pantalla y consultar cualquier multiplicación seleccionando un renglón y una columna. También valida que los números ingresados estén dentro del rango permitido.

```python
# Parte A: Construcción de la matriz

tamaño = 10
matriz = []

for numero_fila in range(1, tamaño + 1):
    nueva_fila = []

    for numero_columna in range(1, tamaño + 1):
        producto = numero_fila * numero_columna
        nueva_fila.append(producto)

    matriz.append(nueva_fila)


# Parte B: Función para imprimir la tabla

def imprimir_tabla(matriz):
    for fila_actual in matriz:
        for valor in fila_actual:
            print(valor, end="\t")
        print()


# Parte C: Función para consultar un producto

def consultar_producto(matriz, renglon, columna):
    return matriz[renglon - 1][columna - 1]


# Parte D: Programa principal

# Mostrar la tabla completa
print("TABLA DE PITÁGORAS")
imprimir_tabla(matriz)

# Solicitar el primer factor
primer_factor = int(input("Ingresa el primer factor: "))

while primer_factor < 1 or primer_factor > 10:
    print("Número fuera de rango. Ingresa un valor del 1 al 10.")
    primer_factor = int(input("Ingresa el primer factor: "))

# Solicitar el segundo factor
segundo_factor = int(input("Ingresa el segundo factor: "))

while segundo_factor < 1 or segundo_factor > 10:
    print("Número fuera de rango. Ingresa un valor del 1 al 10.")
    segundo_factor = int(input("Ingresa el segundo factor: "))

# Consultar y mostrar el resultado
producto_final = consultar_producto(matriz, primer_factor, segundo_factor)

print(primer_factor, "x", segundo_factor, "=", producto_final)
```
